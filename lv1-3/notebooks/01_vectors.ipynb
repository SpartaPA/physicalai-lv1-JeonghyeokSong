## 1-3. 정사영 — 수직성과 합 복원

$\mathbf{a}$ 를 $\mathbf{b}$ 방향으로 정사영한 성분은

$$\mathrm{proj}_{\mathbf{b}}(\mathbf{a})=\frac{\mathbf{a}\cdot\mathbf{b}}{\mathbf{b}\cdot\mathbf{b}}\mathbf{b}$$

이고, 남는 성분(reject)은 $\mathbf{a}-\mathrm{proj}_{\mathbf{b}}(\mathbf{a})$ 입니다.
분모가 $|\mathbf{b}|^2$ 이므로 $\mathbf{b}$ 를 미리 정규화할 필요가 없습니다.

**검증해야 할 두 가지**

1. **수직성**: (남는 성분) $\cdot$ $\mathbf{b} = 0$
2. **합 복원**: $\mathrm{proj} + \mathrm{rej} = \mathbf{a}$

```python
a = np.array([2.0, 3.0, 4.0])
b = np.array([1.0, 0.0, 1.0])

# TODO: project / reject 를 호출하고, 계수 (a·b)/(b·b) 와 함께 결과를 출력하세요.
proj = project(a,b)
print( proj )
print(reject(a, b))

print(f"계수: {dot(a,b)/dot(b,b)}")
```

```python
# --- 검증 ---
rej = reject(a, proj)
coef = dot(a, b) / dot(b, b)

# ① 남는 성분이 b 와 수직인가 (rej·b = 0)
check("① 남는 성분이 b 와 수직인가 (rej·b = 0)", np.isclose(dot(rej, b), 0.0))

# ② proj + rej = a 인가
check("② proj + rej = a 인가", np.allclose(proj + rej, a))

# proj 가 b 와 평행한가 (외적이 0)
check("proj 가 b 와 평행한가 (외적이 0)", np.allclose(cross(proj, b), 0.0))

# 피타고라스: |a|^2 = |proj|^2 + |rej|^2
check("피타고라스: |a|^2 = |proj|^2 + |rej|^2",
      np.isclose(norm(a) ** 2, norm(proj) ** 2 + norm(rej) ** 2))

# 무작위 100 쌍에서도 ①②가 모두 성립하는가  (rng 로 생성)
ok = True
for _ in range(100):
    va = rng.normal(size=3)
    vb = rng.normal(size=3)
    if norm(vb) < 1e-8:
        continue
    vp = project(va, vb)
    vr = reject(va, vp)
    ok &= np.isclose(dot(vr, vb), 0.0) and np.allclose(vp + vr, va)
check("무작위 100 쌍에서도 ①②가 모두 성립하는가", ok)
```

## 1-4. 외적을 반대칭행렬 곱으로 — `skew(a)`

외적은 행렬 곱으로 쓸 수 있습니다.

$$\mathbf{a}\times\mathbf{b} = [\mathbf{a}]_\times \mathbf{b},\qquad
[\mathbf{a}]_\times=\begin{bmatrix}0&-a_3&a_2\a_3&0&-a_1\-a_2&a_1&0\end{bmatrix}$$

이 형태가 중요한 이유는 **로드리게스 공식(문제 2)과 각속도 → 회전 미분**이
전부 $[\boldsymbol{\omega}]_\times$ 로 표현되기 때문입니다.
반대칭(skew-symmetric)이란 $M^{\mathsf{T}} = -M$ 을 뜻합니다.
여기서 따라오는 성질이 하나 더 있는데, 직접 출력해서 확인해 보세요.

```python
a = np.array([1.0, 2.0, 3.0])
b = np.array([4.0, 5.0, 6.0])

# TODO: skew(a) 를 출력하고, skew(a) @ b 와 np.cross(a, b)(# 검산용)를 비교하세요.
print(skew(a), '\n')
# TODO: skew(a).T 와 -skew(a) 를 나란히 출력해 반대칭성을 눈으로 확인하세요.

print(skew(a).T)
print(-skew(a))
print(np.allclose(skew(a).T,-skew(a)))
```

```python
# --- 검증 ---
S = skew(a)

# skew(a) @ b == np.cross(a, b)               (np.cross 는 검산용)
check("skew(a) @ b == np.cross(a, b)", np.allclose(S @ b, np.cross(a, b)))

# skew(a) 가 반대칭인가 (S.T == -S)
check("skew(a) 가 반대칭인가 (S.T == -S)", np.allclose(S.T, -S))

# 반대칭에서 따라오는 대각성분 성질 (M^T = -M 이면 대각성분은 전부 0)
check("반대칭에서 따라오는 대각성분 성질 (대각성분 == 0)", np.allclose(np.diag(S), 0.0))

# skew(a) @ a == 0 (자기 자신과의 외적)
check("skew(a) @ a == 0 (자기 자신과의 외적)", np.allclose(S @ a, 0.0))

# 반교환성: a x b == -(b x a)
check("반교환성: a x b == -(b x a)", np.allclose(cross(a, b), -cross(b, a)))

# 무작위 200 쌍에서 skew 곱 == np.cross
rand_a = rng.normal(size=(200, 3))
rand_b = rng.normal(size=(200, 3))
ok = all(np.allclose(skew(va) @ vb, np.cross(va, vb)) for va, vb in zip(rand_a, rand_b))
check("무작위 200 쌍에서 skew 곱 == np.cross", ok)
```

## 1-5. 세 점이 만드는 평면의 단위 법선

세 점 $P_1,P_2,P_3$ 이 주어지면 두 모서리 벡터
$\mathbf{u}=P_2-P_1$, $\mathbf{v}=P_3-P_1$ 의 외적이 평면에 수직입니다.
이를 정규화하면 단위 법선입니다.

세 점이 **일직선**이면 어떻게 될지 먼저 생각해 보고, 그 경우를 어떻게 처리할지 정하세요.

```python
P1 = np.array([0.0, 0.0, 0.0])
P2 = np.array([1.0, 0.0, 0.0])
P3 = np.array([0.0, 1.0, 0.0])

# TODO: xy 평면 위 세 점의 단위 법선을 구해 출력하고, 기대값과 비교하세요.
print(plane_normal(P1,P2,P3))
# TODO: 기울어진 평면(예: (1,0,0), (0,1,0), (0,0,1))에서도 구해 보세요.
print(plane_normal(np.array([1,0,0]), np.array([0,1,0]), np.array([0,0,1])))
# TODO: 일직선인 세 점을 넣으면 어떻게 되는지 확인해 출력하세요.
print(plane_normal(np.array([1,0,0]),np.array([2,0,0]),np.array([3,0,0])))
```

```python
# --- 검증 ---
n_xy = plane_normal(P1, P2, P3)

# 법선의 길이가 1 인가
check("법선의 길이가 1 인가", np.isclose(norm(n_xy), 1.0))

# xy 평면의 법선이 z축과 일치하는가
check("xy 평면의 법선이 z축과 일치하는가", np.allclose(n_xy, [0.0, 0.0, 1.0]))

# 법선이 두 모서리 벡터 모두와 수직인가
u = P2 - P1
v = P3 - P1
check("법선이 두 모서리 벡터 모두와 수직인가",
      np.isclose(dot(n_xy, u), 0.0) and np.isclose(dot(n_xy, v), 0.0))

# 기울어진 평면의 법선이 기대값과 일치하는가
Q1, Q2, Q3 = np.array([1.0, 0.0, 0.0]), np.array([0.0, 1.0, 0.0]), np.array([0.0, 0.0, 1.0])
n_tilt = plane_normal(Q1, Q2, Q3)
expected = normalize(np.array([1.0, 1.0, 1.0]))     # (1,1,1) 방향, 정규화
check("기울어진 평면의 법선이 기대값과 일치하는가", np.allclose(n_tilt, expected))

# 일직선 입력에서 내가 정한 처리 방식대로 동작하는가 (현재 구현: None 반환)
R1, R2, R3 = np.array([1.0, 0.0, 0.0]), np.array([2.0, 0.0, 0.0]), np.array([3.0, 0.0, 0.0])
n_collinear = plane_normal(R1, R2, R3)
check("일직선 입력에서 내가 정한 처리 방식대로 동작하는가", n_collinear is None)
```

```python
M = np.array([[1.0, 0.0, 1.0],
              [0.0, 1.0, 1.0],
              [1.0, 1.0, 2.0]])

# TODO: row_echelon 으로 사다리꼴과 피벗 열을 출력하세요.
echelon,pivot, _ = row_echelon(M)
print(echelon, pivot)

# TODO: 직접 구현한 rank / det 를 np.linalg.matrix_rank / np.linalg.det (# 검산용) 와 비교하세요.
_rank = np.linalg.matrix_rank(M)
print( np.allclose(_rank, rank(M)) )

_det = np.linalg.det(M)
print( np.allclose(_det, det(M)) )

# TODO: 세 벡터 사이의 선형종속 관계를 코드로 확인해 출력하세요.
#       (스칼라 삼중곱 dot(v1, cross(v2, v3)) 도 같이 보면 좋습니다)
v1, v2, v3 = M[0], M[1], M[2]
print("v3 == v1 + v2:", np.allclose(v3, v1 + v2))
print("스칼라 삼중곱 v1·(v2×v3):", dot(v1, cross(v2, v3)))
```
